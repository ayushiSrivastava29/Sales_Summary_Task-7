Plotting Visuals

```python
import pandas as pd
query = """
SELECT product, SUM(quantity * price) AS total_revenue
FROM SALES
GROUP BY product
"""

df = pd.read_sql(query, conn)
print(df)

# Plot
df.plot(kind='bar', x='product', y='total_revenue', legend=False)
plt.title("Total Revenue by Product")
plt.ylabel("Revenue")
plt.show()
```

```python
query = """
SELECT region, SUM(quantity) AS total_quantity
FROM SALES
GROUP BY region
ORDER BY total_quantity DESC
"""
df = pd.read_sql(query, conn)
print(df)

# Plot pie chart
df.set_index('region').plot(kind='pie', y='total_quantity', autopct='%1.1f%%', legend=False)
plt.title("Total Quantity Sold by Region")
plt.ylabel("") 
plt.show()
```

```python
query = """
SELECT product, SUM(quantity * price) AS total_revenue
FROM sales
GROUP BY product
ORDER BY total_revenue DESC;
"""
df = pd.read_sql(query, conn)
print(df)

# Plot bar chart
df.plot(kind="bar", x="product", y="total_revenue", legend=False)
plt.title("Total Revenue by Product")
plt.ylabel("Revenue")
plt.show()
```

```python
query = """
SELECT TOP 1 product, SUM(quantity) AS total_qty
FROM sales
GROUP BY product
ORDER BY total_qty DESC;
"""
df = pd.read_sql(query, conn)
print(df)

# Plot bar chart
df.plot(kind="bar", x="product", y="total_qty", legend=False)
plt.title("Top Selling Product")
plt.ylabel("Total Quantity Sold")
plt.show()
```
